```python
def area_of_circle(r):
    pi = 3.14159
    area = pi * r * r
    return area

def perimeter_of_circle(r):
    pi = 3.14159
    perimeter = 2 * pi * r
    return perimeter
```

```python
def add_all_nums(*args):
    total = 0
    for num in args:
        if type(num) == int or type(num) == float:
            total = total + num
        else:
            return "Feedback: All items must be numbers"
    return total
```

```python
def convert_celsius_2_fahrenheit(c):
    f = (c * (9/5)) + 32
    return f
```

```python
def check_season(month):
    if month in ["September", "October", "November"]:
        return "Autumn"
    elif month in ["December", "January", "February"]:
        return "Winter"
    elif month in ["March", "April", "May"]:
        return "Spring"
    else:
        return "Summer"
```

```python
def calculate_slope(x1, y1, x2, y2):
    slope = (y2 - y1) / (x2 - x1)
    return slope
```

```python
def solve_quadratic_eqn(a, b, c):
    d = (b * b) - (4 * a * c)
    root1 = (-b + (d ** 0.5)) / (2 * a)
    root2 = (-b - (d ** 0.5)) / (2 * a)
    return root1, root2
```

```python
def print_list(my_list):
    for i in my_list:
        print(i)
```

```python
def capitalize_list_items(my_list):
    new_list = []
    for item in my_list:
        new_list.append(item.upper())
    return new_list
```
